## 2. Service-effectiveness Questions

2a. CA UPT trend by year broken by mode

```python
bus_modes = ["MB", "CB", "RB", "TB"]
rail_modes = ["CR", "HR", "LR", "SR", "YR", "CC", "MG"]

df["mode_group"] = df["mode"].map(
    lambda x: 'Bus' if x in bus_modes else ("Rail" if x in rail_modes else "Other")
)
```

```python
ca_bus_rail_upt =(df
    .groupby(["year", "mode_group"])
    ["unlinked_passenger_trips"]
    .sum()
    .reset_index())
```

```python
ca_bus_rail_upt["unlinked_passenger_trips"] = (ca_bus_rail_upt["unlinked_passenger_trips"].round(0).astype(int))
bus_rail_trend = (ca_bus_rail_upt.pivot(index = "mode_group", columns = "year", values = "unlinked_passenger_trips"))
bus_rail_trend.style.format("{:,}")
```

```python
import altair as alt

alt.data_transformers.enable("vegafusion")

# Select nearest year when hovering
nearest = alt.selection_point(
    nearest=True,
    on="pointerover",
    fields=["year"],
    empty=False
)

# Base line chart
line = alt.Chart(ca_bus_rail_upt).mark_line().encode(
    x="year:O",
    y="unlinked_passenger_trips:Q",
    color="mode_group:N"
)

# Invisible points to capture hover
selectors = alt.Chart(ca_bus_rail_upt).mark_point().encode(
    x="year:O",
    opacity=alt.value(0)
).add_params(
    nearest
)

# Highlight points
points = line.mark_point().encode(
    opacity=alt.when(nearest).then(alt.value(1)).otherwise(alt.value(0)),
    tooltip=[
        "year",
        "mode_group",
        alt.Tooltip("unlinked_passenger_trips:Q", format=",")
    ]
)

# Vertical hover line
rules = alt.Chart(ca_bus_rail_upt).mark_rule(color="gray").encode(
    x="year:O"
).transform_filter(
    nearest
)

# Combine
alt.layer(
    line,
    selectors,
    points,
    rules
).properties(
    width=700,
    height=400
)
```

2c. Do urban transit systems generate more passenger trips per vehicle revenue mile than rural transit systems? (Take 2024 data reference)

```python
df_2024 = (
    df[df["year"] == 2024]
    .groupby("reporting_module")
    .agg(
        vehicle_revenue_miles=("vehicle_revenue_miles", "sum"),
        unique_ntd_ids=("ntd_id", "nunique"),
        avg_upt_per_vrm=("upt_per_vrm", "mean")
    )
    .reset_index()
)

df_2024
```

2d. How many passenger boardings a transit vehicle generates for every hour it is in revenue service? (Across UZAs)

```python
df_CA = df[df["primary_uza_name"].str.endswith(", CA", na=False)]
df_CA["uace_code"] = df_CA["uace_code"].astype(str)
df_CA["uace_code"] = df_CA["uace_code"].str.zfill(5)
df_uza_productivity = (
    df_CA
    .groupby("uace_code")["upt_per_vrh"]
    .mean()
    .reset_index()
)
```

```python
url = "https://www2.census.gov/geo/tiger/TIGER2024/UAC20/tl_2024_us_uac20.zip"
uza_shapes = gpd.read_file(url)
uza_shapes_ca = uza_shapes[uza_shapes["UACE20"].isin(df_CA["uace_code"].unique())]
```

```python
uza_shapes_ca.info()
```

```python
uza_map = uza_shapes_ca.merge(
    df_uza_productivity,
    left_on="UACE20",
    right_on="uace_code",
    how="left"
)
```

## 4. Cost Effectiveness Questions

4a. Which transit modes had the highest farebox recovery ratios in 2024?

```python
farebox_2024 = (
    df[df['year'] == 2024]
    .groupby('mode_full_name')['farebox_recovery_ratio']
    .mean()
    .sort_values(ascending=False)
)

print(farebox_2024)
```

4b. Has farebox recovery improved or declined from 2015 to 2024 across mode groups?

```python
ca_bus_rail_fair_recovery =(df_CA
    .groupby(["year", "mode_group"])
    ["farebox_recovery_ratio"]
    .mean()
    .reset_index())
```

```python
ca_bus_rail_fair_recovery["farebox_recovery_ratio"] = (ca_bus_rail_fair_recovery["farebox_recovery_ratio"].round(3))
bus_rail_trend_fairbox = (ca_bus_rail_fair_recovery.pivot(index = "mode_group", columns = "year", values = "farebox_recovery_ratio"))
bus_rail_trend_fairbox.style.format("{:,}")
```

Farebox recovery declined substantially for both Bus and Rail between 2015 and 2024, while the Other mode group showed a slight overall improvement. Despite a modest recovery after 2021, Bus and Rail farebox recovery ratios remained well below their 2015 levels in 2024. Rail consistently had a higher farebox recovery ratio than Bus.

This calculates the **top 10 most cost-effective agencies in 2024 by ranking agencies with the lowest average net subsidy per passenger trip**, while also summarizing their total ridership and operating costs.


4d. Which agencies are the most cost-effective?

```python
agency_cost_effectiveness = (
    df[df['year'] == 2024]
    .groupby('source_agency')
    .agg(
        net_subsidy_per_trip=('net_subsidy_per_passenger_trip', 'mean'),
        passenger_trips=('unlinked_passenger_trips', 'sum'),
        operating_cost=('operating_expenses_total', 'sum')
    )
    .sort_values('net_subsidy_per_trip')
)

agency_cost_effectiveness.head(10)
```
