```python
import altair as alt

alt.data_transformers.enable("vegafusion")

alt.Chart(
    df[["ntd_id", "mode_full_name", "source_state"]]
).mark_bar().encode(
    x="mode_full_name",
    y="count(ntd_id)",
    column="source_state"
)
```
